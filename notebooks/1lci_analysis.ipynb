```python
import os
import pandas as pd
```

```python
df = pd.read_csv('/home/jupyter-yehlin/ESM3_SaProt_dG/1lci/mutational_scanning_1lci_ensemble.csv')
df_1 = pd.read_csv('/home/jupyter-yehlin/ESM3_SaProt_dG/1lci/mutational_scanning_1lci_model1.csv')
df_2 = pd.read_csv('/home/jupyter-yehlin/ESM3_SaProt_dG/1lci/mutational_scanning_1lci_model2.csv')
df_3 = pd.read_csv('/home/jupyter-yehlin/ESM3_SaProt_dG/1lci/mutational_scanning_1lci_model3.csv')
```

```python
sorted_values = df.iloc[:, 1:].values.flatten()
sorted_indices = sorted(range(len(sorted_values)), key=lambda k: sorted_values[k], reverse=True)
for idx in sorted_indices:
    row = idx // (df.shape[1]-1)
    col = idx % (df.shape[1]-1) + 1
    amino_acid = df.iloc[row, 0]
    value = sorted_values[idx]
    print(f"Position ({amino_acid}, {col}): {value:.3f}")
```

```python
sorted_values = df_1.iloc[:, 1:].values.flatten()
sorted_indices = sorted(range(len(sorted_values)), key=lambda k: sorted_values[k], reverse=True)
for idx in sorted_indices:
    row = idx // (df_1.shape[1]-1)
    col = idx % (df_1.shape[1]-1) + 1
    amino_acid = df_1.iloc[row, 0]
    value = sorted_values[idx]
    print(f"Position ({amino_acid}, {col}): {value:.3f}")
```
